```python
def parse_height(row, col: str = "height") -> float | None:
    value = row[col]
    if isinstance(value, int):
        return float(value)
    if isinstance(value, float):
        return value
    if isinstance(value, str):
        match = re.search(r"([+-]?\d+(?:\.\d+)?)\s*m", value)
        if match:
            return float(match.group(1))
    return np.nan
```

```python
def parse_height_2(row, col: str = "height") -> float | None:
    value = row[col]
    if isinstance(value, int):
        return float(value)
    if isinstance(value, float):
        return value
    if isinstance(value, str):
        match = re.search(r"([+-]?\d+(?:\.\d+)?)\s*m", value)
        if match:
            return float(match.group(1))
    return None
```

```python
buildings_with_height_or_levels["height_corrected"] = (
    buildings_with_height_or_levels.apply(parse_height_2, axis=0)
)
```

```python
import numpy as np

DEFAULT_ROOF_LEVEL_HEIGHT = 3
DEFAULT_LEVEL_HEIGHT = 4
DEFAULT_BUILDING_HEIGHT = 12


def height_calculator(df) -> float:
    height = 0

    # building height
    if df["height"] != np.nan:
        height += df["height"]
    elif df["building:levels"] != np.nan:
        height += df["building:levels"] * DEFAULT_LEVEL_HEIGHT
    if height != 0:  # if we got a height, add a roof height
        if df["roof:height"] != np.nan:
            height += df["roof:height"]
        elif df["roof:levels"] != np.nan:
            height += df["roof:levels"] * DEFAULT_ROOF_LEVEL_HEIGHT
        return height
    return DEFAULT_BUILDING_HEIGHT
```

```python
buildings_with_height_or_levels["adjusted_height"] = (
    buildings_with_height_or_levels.map(height_calculator)
)
```

```python
import pydeck as pdk

# Define the 3D building layer
building_layer = pdk.Layer(
    "PolygonLayer",
    data=buildings_with_height_or_levels,
    get_polygon="geometry.exterior.coords",  # This is where we use the geopandas geometry
    get_elevation="height_c",
    elevation_scale=1,
    get_fill_color=[200, 200, 255, 180],  # Light blue buildings
    pickable=True,
    auto_highlight=True,
    extruded=True,
)

# Set the view state with the initial camera position and pitch for a 3D effect
view_state = pdk.ViewState(
    latitude=VIEW_LATITUDE,
    longitude=VIEW_LONGITUDE,
    zoom=1,  # 14
    pitch=45,
    bearing=0,
)

# Create the Pydeck map
r = pdk.Deck(
    layers=[building_layer],
    initial_view_state=view_state,
    tooltip={"text": "Building Height: {height_c} meters"},
    map_style="mapbox://styles/mapbox/light-v10",  # Requires Mapbox token for non-default styles
)

# You may need a Mapbox API token for non-default map styles.
# Set it as an environment variable (e.g., MAPBOX_API_KEY) or use a free style like:
# map_style='basemaps.cartocdn.com'

# Save the interactive map to an HTML file
output_path = "interactive_3d_map.html"
r.to_html(output_path)

print(f"Interactive 3D map saved to {os.path.abspath(output_path)}")
print("Open the HTML file in your web browser to view the map.")
```
